```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df = pd.read_csv("../data/raw/creditcard.csv")

print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum().sum()
```

```python
fraud_counts = df["Class"].value_counts()
fraud_percentage = df["Class"].value_counts(normalize=True) * 100

print("Fraud Counts:\n", fraud_counts)
print("\nFraud Percentage:\n", fraud_percentage)
```

```python
sns.countplot(x="Class", data=df)
plt.title("Fraud vs Non-Fraud Distribution")
plt.xticks([0,1], ["Legitimate", "Fraud"])
plt.show()
```

```python
plt.figure()
sns.boxplot(x="Class", y="Amount", data=df)
plt.yscale("log")
plt.title("Transaction Amount Distribution by Class (Log Scale)")
plt.show()
```

```python
df.groupby("Class")["Amount"].describe()
```

```python
sns.histplot(data=df, x="Amount", hue="Class", bins=50, log_scale=True)
plt.title("Amount Distribution (Log Scale)")
plt.show()
```

```python
plt.figure()
sns.boxplot(x="Class", y="Amount", data=df)
plt.yscale("log")
plt.title("Transaction Amount Distribution by Class (Log Scale)")
plt.show()
```

```python
df.groupby("Class")["Amount"].describe()
```

```python
sns.histplot(data=df, x="Amount", hue="Class", bins=50, log_scale=True)
plt.title("Amount Distribution (Log Scale)")
plt.show()
```

```python
df["Hour"] = (df["Time"] // 3600) % 24
df["Hour"] = df["Hour"].astype(int)

df.head()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x="Hour", hue="Class", data=df)
plt.title("Fraud Distribution by Hour of Day")
plt.show()
```

```python
fraud_by_hour = df.groupby("Hour")["Class"].mean()

fraud_by_hour.plot(kind="bar")
plt.title("Fraud Rate by Hour")
plt.ylabel("Fraud Rate")
plt.show()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(df.corr(), cmap="coolwarm", center=0)
plt.title("Correlation Matrix")
plt.show()
```
